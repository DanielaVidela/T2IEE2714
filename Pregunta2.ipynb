# Pregunta 2 — Difusión anisotrópica, Variación Total y diseño del coeficiente de difusión

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

cargo primero la imagen


```python
img = cv2.imread('P4_IMG_2267_CFA.tif', cv2.IMREAD_GRAYSCALE)

#normalizar la imágen
imgprueba = img.astype(np.float64)
imgprueba = (imgprueba - np.min(imgprueba)) / (np.max(imgprueba) - np.min(imgprueba))

#ver que se haya cargado correctamente
plt.imshow(imgprueba, cmap='gray', vmin=0, vmax=1)
plt.show()
```

```python
#repito pasos de la p1
np.random.seed(1)
ruido = np.random.normal(loc=0.0, scale=0.05, size=imgprueba.shape)

imgruidosa = imgprueba + ruido

imgruidosa = np.clip(imgruidosa, 0, 1)

plt.imshow(imgruidosa, cmap='gray', vmin=0, vmax=1)
plt.show()
```
